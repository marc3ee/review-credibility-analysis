```python
import pandas as pd
import re

# Load the lexicon file
lexicon_df = pd.read_excel("PH Word Sentiments Full.xlsx", engine="openpyxl")
lexicon_df = lexicon_df[['word', 'sentiment']].dropna()
lexicon_df['sentiment_score'] = lexicon_df['sentiment'].map({'positive': 1, 'negative': -1})
lexicon_dict = dict(zip(lexicon_df['word'], lexicon_df['sentiment_score']))

# Load your review data
reviews_df = pd.read_csv("reviews_2025-04-17_22-15-40.csv")

# Define sentiment classification function
def classify_review_sentiment(text, lexicon):
    words = re.findall(r'\b\w+\b', str(text).lower())
    score = sum(lexicon.get(word, 0) for word in words)
    if score > 0:
        return 'positive'
    elif score < 0:
        return 'negative'
    else:
        return 'neutral'

# Apply to each review
reviews_df['Lexicon Sentiment'] = reviews_df['Review'].apply(lambda x: classify_review_sentiment(x, lexicon_dict))

# Preview result
print(reviews_df[['Review', 'Lexicon Sentiment']].head(10))
```

```python
reviews_df.to_excel("classified_reviews.xlsx", index=False, engine="openpyxl")
```
